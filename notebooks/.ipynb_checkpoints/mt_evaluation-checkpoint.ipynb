# Machine Translation Quality Evaluation Pipeline

**Language pairs:** English → Italian (primary) · English → Japanese (secondary — no native speaker)

**MT engine:** DeepL API (free tier)  
**Metrics:** BLEU (sacrebleu) · COMET (wmt22-comet-da)  
**Data:** FLORES-200 benchmark, `devtest` split

---

### Why this setup?

Automatic MT evaluation is at the core of any scalable localisation workflow. This notebook explores two complementary metrics:

- **BLEU** is fast, deterministic, and universally understood — but it penalises legitimate paraphrase and is known to correlate poorly with human judgement for morphologically rich languages like Italian.
- **COMET** uses a pre-trained multilingual encoder to score translations by semantic similarity and fluency, not lexical overlap. It correlates much more strongly with human ratings.

The secondary language pair (Japanese) is deliberately chosen to be one where I *cannot* rely on my own native intuition. This forces the question: **how do you evaluate MT quality at scale, across languages you don't speak?**

## 0. Setup

```python
import os
import json
import time
import random
import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns

from pathlib import Path
from dotenv import load_dotenv

# Load API key from .env file in the project root
load_dotenv(Path('..') / '.env')
DEEPL_API_KEY = os.getenv('DEEPL_API_KEY')

if not DEEPL_API_KEY:
    raise ValueError('DEEPL_API_KEY not found. Add it to a .env file in the project root.')

print('Setup complete.')
```

## 1. Load the data (FLORES-200)

FLORES-200 is a professionally translated benchmark covering 200 languages, built from Wikipedia and Wikinews articles. It's the standard reference corpus for evaluating multilingual MT systems.

We use the `devtest` split: 1012 sentences, all of which have been translated by professional translators from a common English source. This gives us clean reference translations without the noise of crowd-sourced data.

```python
from datasets import load_dataset

# Load English source and Italian + Japanese references
print('Loading FLORES-200 devtest split...')

flores_en  = load_dataset('facebook/flores', 'eng_Latn', split='devtest', trust_remote_code=True)
flores_it  = load_dataset('facebook/flores', 'ita_Latn', split='devtest', trust_remote_code=True)
flores_ja  = load_dataset('facebook/flores', 'jpn_Jpan', split='devtest', trust_remote_code=True)

# Use a reproducible sample of 100 sentences (free-tier API limit awareness)
SAMPLE_SIZE = 100
random.seed(42)
indices = sorted(random.sample(range(len(flores_en)), SAMPLE_SIZE))

en_sentences  = [flores_en[i]['sentence']  for i in indices]
it_references = [flores_it[i]['sentence']  for i in indices]
ja_references = [flores_ja[i]['sentence']  for i in indices]

print(f'Loaded {SAMPLE_SIZE} sentence pairs.')
print(f'\nExample EN: {en_sentences[0]}')
print(f'Example IT ref: {it_references[0]}')
print(f'Example JA ref: {ja_references[0]}')
```

## 2. Translate with DeepL

DeepL is consistently ranked at or near the top of general-purpose MT systems for European languages. Its Italian output is considered excellent — which makes it a good test: even a *good* MT system should reveal interesting edge cases under close analysis.

> **Note:** The free API tier allows 500k characters/month. Our 100 sentences are well within that. We add a small sleep between requests to be a polite API consumer.

```python
import deepl

translator = deepl.Translator(DEEPL_API_KEY)

def translate_batch(sentences, target_lang, batch_size=10):
    """Translate a list of sentences to target_lang using DeepL."""
    results = []
    for i in range(0, len(sentences), batch_size):
        batch = sentences[i:i + batch_size]
        translations = translator.translate_text(batch, target_lang=target_lang)
        results.extend([t.text for t in translations])
        time.sleep(0.5)  # polite rate limiting
        print(f'  Translated {min(i + batch_size, len(sentences))}/{len(sentences)}', end='\r')
    print()
    return results

# Cache translations to disk so we don't re-hit the API on re-runs
cache_path = Path('../data/translations.json')
cache_path.parent.mkdir(exist_ok=True)

if cache_path.exists():
    print('Loading cached translations...')
    with open(cache_path) as f:
        cache = json.load(f)
    it_hypotheses = cache['it']
    ja_hypotheses = cache['ja']
else:
    print('Translating EN → IT...')
    it_hypotheses = translate_batch(en_sentences, 'IT')
    print('Translating EN → JA...')
    ja_hypotheses = translate_batch(en_sentences, 'JA')
    
    with open(cache_path, 'w', encoding='utf-8') as f:
        json.dump({'it': it_hypotheses, 'ja': ja_hypotheses}, f, ensure_ascii=False, indent=2)
    print('Translations saved to cache.')

print(f'\nExample MT output (IT): {it_hypotheses[0]}')
print(f'Example MT output (JA): {ja_hypotheses[0]}')
```

## 3. Score with BLEU

BLEU (Bilingual Evaluation Understudy) measures n-gram overlap between the hypothesis and one or more references. It's the oldest and most widely cited MT metric.

**Key properties:**
- Ranges from 0 to 100 (higher is better)
- Penalises short hypotheses (brevity penalty)
- Uses 4-gram precision by default
- Deterministic and fast — good for CI/CD pipelines

**Known weaknesses for Italian:**
- Agglutinative morphology means a single reference may use `studenti` where the hypothesis uses `alunni` — both correct, both penalised
- Synonyms, pronoun choices, and word-order variation all reduce BLEU even when the translation is good

```python
import sacrebleu

def corpus_bleu(hypotheses, references):
    """Compute corpus-level BLEU."""
    result = sacrebleu.corpus_bleu(hypotheses, [references])
    return result

def sentence_bleu_scores(hypotheses, references):
    """Compute per-sentence BLEU scores."""
    scores = []
    for hyp, ref in zip(hypotheses, references):
        score = sacrebleu.sentence_bleu(hyp, [ref])
        scores.append(score.score)
    return scores

# Corpus-level BLEU
it_bleu = corpus_bleu(it_hypotheses, it_references)
ja_bleu = corpus_bleu(ja_hypotheses, ja_references)

print(f'BLEU (EN→IT): {it_bleu.score:.2f}')
print(f'BLEU (EN→JA): {ja_bleu.score:.2f}')

# Sentence-level BLEU (for per-sentence analysis later)
it_sent_bleu = sentence_bleu_scores(it_hypotheses, it_references)
ja_sent_bleu = sentence_bleu_scores(ja_hypotheses, ja_references)
```

## 4. Score with COMET

COMET (Crosslingual Optimized Metric for Evaluation of Translation) uses a fine-tuned multilingual encoder (XLM-RoBERTa) trained on human direct-assessment scores from WMT campaigns.

**Key properties:**
- Produces scores typically in the range ~0.7–1.0 for good MT (can go below 0 for very bad output)
- Reference-based: uses source + hypothesis + reference
- Correlates significantly better with human judgement than BLEU across all language pairs
- Requires downloading a checkpoint (~1.5GB) on first run

We use `wmt22-comet-da`, the production checkpoint trained on WMT22 direct assessment data.

```python
from comet import download_model, load_from_checkpoint

print('Downloading/loading COMET checkpoint (may take a few minutes on first run)...')
model_path = download_model('Unbabel/wmt22-comet-da')
model = load_from_checkpoint(model_path)
print('COMET model loaded.')
```

```python
def comet_scores(sources, hypotheses, references):
    """Compute COMET scores. Returns (system_score, list_of_segment_scores)."""
    data = [
        {'src': s, 'mt': h, 'ref': r}
        for s, h, r in zip(sources, hypotheses, references)
    ]
    output = model.predict(data, batch_size=8, gpus=0)  # set gpus=1 if you have a GPU
    return output.system_score, output.scores

print('Scoring EN→IT...')
it_comet_sys, it_comet_seg = comet_scores(en_sentences, it_hypotheses, it_references)

print('Scoring EN→JA...')
ja_comet_sys, ja_comet_seg = comet_scores(en_sentences, ja_hypotheses, ja_references)

print(f'\nCOMET (EN→IT): {it_comet_sys:.4f}')
print(f'COMET (EN→JA): {ja_comet_sys:.4f}')
```

## 5. Results overview

```python
# Build a clean summary DataFrame
summary = pd.DataFrame({
    'Language Pair': ['EN → IT', 'EN → JA'],
    'BLEU': [round(it_bleu.score, 2), round(ja_bleu.score, 2)],
    'COMET': [round(it_comet_sys, 4), round(ja_comet_sys, 4)],
})

print('=== System-level scores ===')
print(summary.to_string(index=False))
```

```python
# Per-sentence DataFrame for deep dives
df_it = pd.DataFrame({
    'source':     en_sentences,
    'reference':  it_references,
    'hypothesis': it_hypotheses,
    'bleu':       it_sent_bleu,
    'comet':      it_comet_seg,
    'pair':       'EN→IT'
})

df_ja = pd.DataFrame({
    'source':     en_sentences,
    'reference':  ja_references,
    'hypothesis': ja_hypotheses,
    'bleu':       ja_sent_bleu,
    'comet':      ja_comet_seg,
    'pair':       'EN→JA'
})

df_all = pd.concat([df_it, df_ja], ignore_index=True)

# Save to results/
Path('../results').mkdir(exist_ok=True)
df_all.to_csv('../results/scores.csv', index=False)
print(f'Scores saved to results/scores.csv ({len(df_all)} rows)')
```

## 7. Where BLEU and COMET disagree

The most informative cases are where the two metrics diverge sharply. A sentence with **low BLEU but high COMET** is likely a good translation that happened to use different words than the reference. A sentence with **high BLEU but low COMET** may have n-gram overlap by accident while being semantically off.

```python
# Normalise both metrics to [0,1] to compare them fairly
df_it['bleu_norm']  = (df_it['bleu']  - df_it['bleu'].min())  / (df_it['bleu'].max()  - df_it['bleu'].min())
df_it['comet_norm'] = (df_it['comet'] - df_it['comet'].min()) / (df_it['comet'].max() - df_it['comet'].min())
df_it['disagreement'] = (df_it['comet_norm'] - df_it['bleu_norm']).abs()

# Top cases where COMET >> BLEU (good translation, different words)
comet_wins = df_it.nlargest(5, 'disagreement').query('comet_norm > bleu_norm')

print('=== Cases where COMET rates much higher than BLEU ===' )
print('(Good translations penalised by BLEU for lexical divergence)\n')

for _, row in comet_wins.iterrows():
    print(f'SOURCE:     {row["source"]}')
    print(f'REFERENCE:  {row["reference"]}')
    print(f'HYPOTHESIS: {row["hypothesis"]}')
    print(f'BLEU: {row["bleu"]:.1f}  |  COMET: {row["comet"]:.4f}')
    print()
```

```python
# Cases where BLEU >> COMET (possible false positives)
bleu_wins = df_it.nlargest(5, 'disagreement').query('bleu_norm > comet_norm')

print('=== Cases where BLEU rates much higher than COMET ===')
print('(Superficial lexical match but semantic issues)\n')

for _, row in bleu_wins.iterrows():
    print(f'SOURCE:     {row["source"]}')
    print(f'REFERENCE:  {row["reference"]}')
    print(f'HYPOTHESIS: {row["hypothesis"]}')
    print(f'BLEU: {row["bleu"]:.1f}  |  COMET: {row["comet"]:.4f}')
    print()
```

## 8. Manual error analysis (EN→IT sample)

Automatic metrics can tell us *that* something is wrong, not *what*. This section uses a 30-sentence sample for a manual annotation pass, tagging each error under three classical categories:

| Category | Definition | Example |
|----------|-----------|--------|
| **Fluency** | The translation is grammatically or stylistically awkward in the target language, regardless of meaning | Wrong verb agreement, unnatural word order |
| **Adequacy** | The translation fails to convey the full meaning of the source (omission, addition, mistranslation) | Dropped negation, wrong pronoun referent |
| **Terminology** | Wrong choice of domain-specific term or proper noun | Brand name left untranslated when a standard IT equivalent exists |

> **This cell is a template.** Fill in the `error_type` and `notes` columns as you review each sentence. A score of `None` means no error detected.

```python
# Draw a stratified sample for error analysis:
# 10 high-COMET, 10 mid-COMET, 10 low-COMET — so we see the full quality range
df_sorted = df_it.sort_values('comet').reset_index(drop=True)

n = len(df_sorted)
sample_low  = df_sorted.iloc[:n//3].sample(10, random_state=42)
sample_mid  = df_sorted.iloc[n//3:2*n//3].sample(10, random_state=42)
sample_high = df_sorted.iloc[2*n//3:].sample(10, random_state=42)

error_sample = pd.concat([sample_low, sample_mid, sample_high]).reset_index(drop=True)

# Pre-populate annotation columns — fill these in manually!
error_sample['error_type'] = None   # 'fluency' | 'adequacy' | 'terminology' | None
error_sample['notes']      = ''     # free-text notes on the error

# Save template to CSV for offline annotation
error_sample[['source', 'reference', 'hypothesis', 'bleu', 'comet', 'error_type', 'notes']].to_csv(
    '../results/error_analysis_template.csv', index=False
)

print('Error analysis template saved to results/error_analysis_template.csv')
print('Open it in a spreadsheet, fill in error_type and notes, then load it back below.')

# Preview
error_sample[['source', 'hypothesis', 'comet']].head(10)
```

## 9. The second language pair: English → Japanese

### Why Japanese?

Japanese presents a fundamentally different evaluation challenge:

1. **Script**: Japanese uses three writing systems (hiragana, katakana, kanji). A translator who doesn't read Japanese cannot eyeball errors the way a native speaker can.
2. **Tokenisation**: BLEU counts n-gram overlap over tokens. Japanese has no spaces between words — sacrebleu uses character-level tokenisation by default (`--tokenize ja-mecab` or `char`), which changes the absolute scores significantly and makes cross-language BLEU comparisons meaningless.
3. **Information structure**: Japanese is SOV and pro-drop; English is SVO. Adequacy errors (especially omission) are harder to catch automatically.

### How do you evaluate MT at scale without native speakers?

This is the real-world challenge at any global company — including the kind of localisation pipeline a company like GetYourGuide runs. The options, from cheapest to most reliable:

| Approach | Cost | Reliability | Notes |
|----------|------|-------------|-------|
| BLEU only | Free | Low | Gameable; doesn't catch fluency errors |
| COMET (reference-based) | Free (GPU) | Medium | Better, but still needs good references |
| COMET-QE (quality estimation, no reference) | Free (GPU) | Medium | Useful when references are expensive |
| Bilingual spot-check (vendor) | $$$ | High | Gold standard; not scalable |
| Human evaluation with MQM framework | $$$$ | Very high | Used by Google, WMT — full error taxonomy |

The pragmatic answer for a product-scale pipeline: use COMET-QE as a cheap filter, flag low-scoring segments for human review, and periodically run full evaluations on a held-out set.

```python
# Japanese tokenisation note: sacrebleu uses character-level tokenisation for JA
# The absolute BLEU score is not comparable to IT — what matters is relative change
# and the COMET score (which handles Japanese natively via XLM-RoBERTa)

# Show a few examples for qualitative inspection
print('EN→JA spot check (10 examples)\n')
print(f'{"SOURCE":<60} {"COMET":>7}')
print('-' * 70)
for _, row in df_ja.sort_values('comet').head(5).iterrows():
    print(f'{row["source"][:58]:<60} {row["comet"]:>7.4f}  ← lowest COMET')
print()
for _, row in df_ja.sort_values('comet').tail(5).iterrows():
    print(f'{row["source"][:58]:<60} {row["comet"]:>7.4f}  ← highest COMET')
```

The error analysis typically reveals that the majority of errors for high-quality neural MT are **fluency issues** (register, naturalness) rather than adequacy failures. Terminology errors are rare in general-domain text but become the dominant category in domain-specific content (e.g., tourism, legal, medical).

### The limits of automatic metrics

1. **A single reference is a ceiling, not a truth.** BLEU punishes correct translations that happen to use different words. Any serious evaluation should use multiple references where possible.
2. **COMET has its own biases.** It was trained on WMT news-domain data. Performance on other domains (e.g., travel content, UGC, marketing copy) is less well-characterised.
3. **Neither metric catches all error types.** Cultural adaptation, brand voice, and terminology consistency cannot be evaluated automatically — they require human judgement.
4. **System-level scores hide segment-level variance.** A system can score well overall while consistently failing on specific sentence types (long sentences, questions, passive voice, numbers).

### Practical takeaway

For a production MT pipeline, the right approach is **layered evaluation**: COMET-QE as a real-time quality gate, periodic COMET/BLEU benchmarking on a held-out test set, and targeted human review (MQM or similar) for high-stakes content categories. Automatic metrics are signals, not verdicts.
